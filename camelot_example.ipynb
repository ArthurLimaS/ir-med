# Example of csv_creator usage

### Read tables

```python
import camelot
import etl_functions as etl

EDITAL = '1616_2023_14'

tables = camelot.read_pdf(
    'data/notices/pdfs/LICON_Edital_1616_2023_14_1333550.pdf',
    pages='39-43'
)
tables
```

### Validate Consistency of Extracted Tables

Check whether all extracted tables have the same number of rows.  
This is a quick way to verify if the table extraction process was successful.

```python
count = 0
for tb in tables:
    n_cols = len(tb.df.columns)

    print(f"{count} - {n_cols}")
    count += 1
```

```python
NEW_HEADER = ['_', '_', 'catmat', 'descricao', 'und', 'quant', 'v_unit',
              'v_total']

etl.csv_creator(
    tables,
    f"data/notices/csvs/LE_{EDITAL}_new.csv",
    header = NEW_HEADER,
    drop_columns = [0, 1],
    descricao_column = 1,
    preco_column = 4
)
```
